```python

from collections import OrderedDict

class MapReduce:

    def __init__(self):
        self.intermediate = OrderedDict()
        self.result = []

    def emitIntermediate(self, key, value):
        self.intermediate.setdefault(key, [])
        self.intermediate[key].append(value)

    def emit(self, value):
        i, j, val = value
        self.result[i][j] = val

    def execute(self, matrix1, matrix2, mapper, reducer):
        n = len(matrix1)
        m = len(matrix2[0])

        self.result = [[0] * m for _ in range(n)]

        mapper(matrix1, matrix2)

        for key in self.intermediate:
            reducer(key, self.intermediate[key])

        print("\nResult Matrix:")
        for row in self.result:
            print(*row)


mapReducer = None


def mapper(matrix1, matrix2):
    n = len(matrix1)
    rows_b = len(matrix2)
    cols_b = len(matrix2[0])

    for i in range(n):
        for j in range(cols_b):

            mapReducer.emitIntermediate((i, j), matrix1[i])

            column = []
            for k in range(rows_b):
                column.append(matrix2[k][j])

            mapReducer.emitIntermediate((i, j), column)


def reducer(key, list_of_values):
    row = list_of_values[0]
    column = list_of_values[1]

    dot_product = 0

    for i in range(len(row)):
        dot_product += row[i] * column[i]

    mapReducer.emit((key[0], key[1], dot_product))


if __name__ == "__main__":

    testcases = int(input("Enter number of test cases: "))

    for _ in range(testcases):

        mapReducer = MapReduce()

        rows1, cols1 = map(
            int, input("Enter rows and columns of Matrix A: ").split()
        )

        matrix1 = []
        print("Enter Matrix A:")
        for _ in range(rows1):
            matrix1.append(list(map(int, input().split())))

        rows2, cols2 = map(
            int, input("Enter rows and columns of Matrix B: ").split()
        )

        matrix2 = []
        print("Enter Matrix B:")
        for _ in range(rows2):
            matrix2.append(list(map(int, input().split())))

        if cols1 != rows2:
            print("Matrix multiplication not possible!")
        else:
            mapReducer.execute(matrix1, matrix2, mapper, reducer)

'''
Enter number of test cases: 1
Enter rows and columns of Matrix A: 3 2
Enter Matrix A:
5 6
8 7
2 4
Enter rows and columns of Matrix B: 2 3
Enter Matrix B:
1 2 6
7 5 4

Result Matrix:
47 40 54
57 51 76
30 24 28

'''
```
