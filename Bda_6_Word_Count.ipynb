```python
from collections import defaultdict
text = input("Enter the text: ")

# Map phase
def mapper(text):
  return [(word, 1) for word in text.split()]
mapped = mapper(text)

# Shuffle phase
shuffled = defaultdict(list)
for word, count in mapped:
  shuffled[word].append(count)

# Reduce phase
def reducer(shuffled):
  return {word: sum(counts) for word, counts in shuffled.items()}
result = reducer(shuffled)
print("\nWord Count:")
for word, count in result.items():
  print(word, ":", count)

# Enter the text: I love pizza , I love coke , I love sports , I am Student , Who am I actually ?
```
