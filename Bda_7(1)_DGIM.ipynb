```python

class DGIM:
    def __init__(self, window_size):
        self.window_size = window_size
        self.buckets = []
        self.timestamp = 0

    def add_bit(self, bit):
        self.timestamp += 1

        if bit == 1:
            self.buckets.append([1, self.timestamp])
            self.merge()

        # Remove old buckets
        self.buckets = [
            b for b in self.buckets
            if b[1] > self.timestamp - self.window_size
        ]

    def merge(self):
        while True:
            merged = False

            for size in [1, 2, 4, 8, 16, 32, 64]:
                indexes = [
                    i for i, b in enumerate(self.buckets)
                    if b[0] == size
                ]

                if len(indexes) > 2:
                    i1 = indexes[0]
                    i2 = indexes[1]

                    new_bucket = [
                        size * 2,
                        self.buckets[i2][1]
                    ]

                    self.buckets.pop(i2)
                    self.buckets.pop(i1)
                    self.buckets.insert(i1, new_bucket)

                    merged = True
                    break

            if not merged:
                break

    def query(self):
        total = 0

        for bucket in self.buckets:
            total += bucket[0]

        return total


# Main Program

window = int(input("Enter window size: "))
stream = input("Enter binary stream: ")

dgim = DGIM(window)

for bit in stream:
    if bit in "01":
        dgim.add_bit(int(bit))

print("\nDGIM Buckets:")

for bucket in reversed(dgim.buckets):
    print("Size of bucket:", bucket[0],
          ", Timestamp:", bucket[1])

print("\nEstimated number of ones in the last",
      window, "bits:", dgim.query())


'''
Enter window size: 10
Enter binary stream: 10110110101
'''


'''
# DGIM Algorithm

DGIM (Datar-Gionis-Indyk-Motwani) Algorithm is a data stream algorithm used to estimate the number of 1s in the last N bits of a binary data stream. It is mainly used when data arrives continuously and storing the entire data stream is not practical.

DGIM uses a bucket-based approach to reduce memory usage. Each bucket stores two values: the size of the bucket and the timestamp of its most recent bit. Bucket sizes are always maintained as powers of 2, such as 1, 2, 4, 8, and 16.

### Working of DGIM Algorithm

1. New Bit Arrival: Whenever a new bit arrives, the timestamp is increased.
2. Bucket Creation: If the incoming bit is 1, a new bucket of size 1 is created.
3. Bucket Merging: If more than two buckets have the same size, the two oldest buckets are merged into one bucket of double size.
4. Window Maintenance: Buckets that fall outside the required window are removed.
5. Query Processing: The algorithm estimates the number of 1s by adding bucket sizes and approximately counting half of the oldest bucket.

### Advantages

- Requires less memory than storing the complete data stream.
- Processes incoming data efficiently.
- Useful for large and continuous data streams.
- Provides approximate answers with controlled error.

### Applications

DGIM is used in network traffic monitoring, web analytics, sensor data analysis, and real-time data stream processing.

Conclusion: The DGIM algorithm is an efficient technique for estimating the number of 1s in a recent binary data stream using buckets. It reduces memory requirements while providing approximate results.
'''
```
