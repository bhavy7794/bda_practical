```python

import hashlib
import math

class BloomFilter:

    def __init__(self, size, hash_count):
        self.size = size
        self.hash_count = hash_count
        self.bit_array = [0] * size

    def _hashes(self, item):
        indexes = []

        for i in range(self.hash_count):
            data = str(item) + str(i)

            hash_value = int(
                hashlib.md5(data.encode()).hexdigest(), 16
            )

            indexes.append(hash_value % self.size)

        return indexes

    def add(self, item):
        for index in self._hashes(item):
            self.bit_array[index] = 1

    def contains(self, item):
        for index in self._hashes(item):
            if self.bit_array[index] == 0:
                return False

        return True


# Main Program

size = int(input("Enter size of bit array: "))
hash_count = int(input("Enter number of hash functions: "))

bloom = BloomFilter(size, hash_count)

n = int(input("Enter number of elements: "))

print("\nEnter elements:")

for i in range(n):
    word = input()
    bloom.add(word)


# False Positive Probability

probability = (1 - math.exp(
    -(hash_count * n) / size
)) ** hash_count

print("\nSize of bit array:", size)

print("False Positive Probability:",
      round(probability, 3))

print("Number of hash functions:", hash_count)


# Searching Elements

search_count = int(input("\nEnter number of elements to search: "))

print("\nSearch Results:")

for i in range(search_count):
    word = input("Enter word to search: ")

    if bloom.contains(word):
        print("'" + word + "' is probably present!")
    else:
        print("'" + word + "' is definitely not present!")


print("\n \n \n \n")
'''
Enter size of bit array: 20
Enter number of hash functions: 3c

Enter number of elements: 3

Enter elements:
apple
banana
mango

Enter number of elements to search: 3

Enter word to search: apple
Enter word to search: grapes
Enter word to search: mango
'''




'''
Bloom Filter Algorithm
Bloom Filter is a space-efficient probabilistic data structure used to check whether an element is present in a set or not. It is commonly used in Big Data applications where storing large amounts of data requires more memory.
A Bloom Filter uses a bit array and multiple hash functions to store and check elements. Initially, all bits in the array are set to 0. When an element is inserted, it is passed through multiple hash functions, which generate positions in the bit array. The bits at those positions are changed to 1.

Working of Bloom Filter

1. Initialization: Create a bit array of a fixed size with all bits set to 0.
2. Insertion: Apply multiple hash functions to the element and set the generated bit positions to 1.
3. Searching: Apply the same hash functions to the search element.
4. If any generated position contains 0, the element is definitely not present.
5. If all generated positions contain 1, the element is probably present.

Advantages

- Requires less memory.
- Provides fast insertion and searching.
- Efficient for handling large datasets.
- Does not produce false negatives in a standard Bloom Filter.

Applications

Bloom Filters are used in database systems, web caching, network security, duplicate detection, and search engines.

Conclusion: Bloom Filter is an efficient probabilistic technique used to test the membership of elements. It saves memory and provides fast searching, but it may produce false positives, where an element is reported as probably present even when it is absent.
'''
```
